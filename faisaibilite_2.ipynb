# <a id="1"></a> 1. Text Data Processing: Word/Sentence Embedding

Word/sentence embedding is a natural language processing technique that allows words and sentences to be represented as numerical vectors. It assigns a vector to each word or sentence that captures semantic and syntactic characteristics of the language.

## <a id="2"></a> 1.1 Word2Vec

Word2Vec is a machine learning algorithm for creating word embeddings. It is based on an artificial neural network that learns to predict the neighboring words of a given word in a text corpus.


```python
descriptions = df_cat['descr_bow_lem'].to_list()
descriptions = [gs.utils.simple_preprocess(text) for text in descriptions]
```

```python
print("max length bow lem : ", df_cat['length_bow_lem'].max())
```

```python
# Déterminer les paramètres
w2v_size=300
w2v_window=5
w2v_min_count=1
w2v_epochs=100
maxlen = 380
```

```python
print("Création et entraînement du modèle Word2Vec.")
w2v_model = gs.models.Word2Vec(min_count=w2v_min_count,
                               window=w2v_window,
                               vector_size=w2v_size,
                               seed=0)


w2v_model.build_vocab(descriptions)
w2v_model.train(descriptions, total_examples=w2v_model.corpus_count, epochs=w2v_epochs)
w2v_words = w2v_model.wv.index_to_key
print("Taille du vocabulaire :", len(w2v_words))
print("Word2Vec entrainé.")
```

```python
# Export du modèle
with open('models/word2vec', 'wb') as f1:
    pickle.dump(w2v_model, f1)
```

```python
print("Préparation des descriptions")
tokenizer = tf.preprocessing.text.Tokenizer()
tokenizer.fit_on_texts(descriptions)
x_sentences = tf.preprocessing.sequence.pad_sequences(tokenizer.texts_to_sequences(descriptions),
                                                     maxlen=maxlen,
                                                     padding='post') 
                                                   
num_words = len(tokenizer.word_index) + 1
print("Number of unique words: ", num_words)
```

```python
print("Création de la matrice d'embedding")
word_index = tokenizer.word_index
vocab_size = len(word_index) + 1
embedding_matrix = np.zeros((vocab_size, w2v_size))
i=0
j=0
    
for word, idx in word_index.items():
    i +=1
    if word in w2v_words:
        j +=1
        embedding_vector = w2v_model.wv[word]
        if embedding_vector is not None:
            embedding_matrix[idx] = w2v_model.wv[word]
            
word_rate = np.round(j/i,4)
print("Word embedding rate : ", word_rate)
print("Embedding matrix: %s" % str(embedding_matrix.shape))
```

```python
# Création du modèle

input=tf.layers.Input(shape=(len(x_sentences),maxlen),dtype='float64')
word_input=tf.layers.Input(shape=(maxlen,),dtype='float64')  
word_embedding=tf.layers.Embedding(input_dim=vocab_size,
                                   output_dim=w2v_size,
                                   weights = [embedding_matrix],
                                   input_length=maxlen)(word_input)
word_vec=tf.layers.GlobalAveragePooling1D()(word_embedding)  
embed_model = tf.models.Model([word_input],word_vec)

embed_model.summary()
```

* 0 indicates that the partitions are independent.
* Negative values indicate that the partitions are less similar than what would be expected by chance.

## <a id="3"></a> 1.2 BERT HuggingFace

BERT (Bidirectional Encoder Representations from Transformers) is a natural language processing (NLP) model based on deep learning. It is a pre-trained language representation model trained on vast amounts of unlabeled text data.

BERT uses a bidirectional transformer neural network to capture the relationships between words in a sentence.

The BERT model has been pre-trained on gigabytes of text data from various sources such as Wikipedia and Common Crawl. It can then be fine-tuned for specific tasks such as text classification, sentiment analysis, machine translation, etc.

The Hugging Face BERT model is a pre-trained implementation of BERT that can be used for NLP tasks.

'bert-base-uncased' is one of the pre-trained BERT models. It is a base version of the BERT model that has been trained on a large text corpus and can be used for specific NLP tasks such as text classification, sentiment analysis, machine translation, etc.


```python
def bert_inp_fct(descriptions, bert_tokenizer) :
    '''Fonction de préparation des descriptions.'''
    input_ids=[]
    token_type_ids = []
    attention_mask=[]
    bert_inp_tot = []

    for desc in descriptions:
        bert_inp = bert_tokenizer.encode_plus(desc,
                                              add_special_tokens = True,
                                              #max_length = max_length,
                                              padding='max_length',
                                              return_attention_mask = True, 
                                              return_token_type_ids=True,
                                              truncation=True,
                                              return_tensors="tf")
    
        input_ids.append(bert_inp['input_ids'][0])
        token_type_ids.append(bert_inp['token_type_ids'][0])
        attention_mask.append(bert_inp['attention_mask'][0])
        bert_inp_tot.append((bert_inp['input_ids'][0], 
                             bert_inp['token_type_ids'][0], 
                             bert_inp['attention_mask'][0]))

    input_ids = np.asarray(input_ids)
    token_type_ids = np.asarray(token_type_ids)
    attention_mask = np.array(attention_mask)
    
    return input_ids, token_type_ids, attention_mask, bert_inp_tot
    

def feature_BERT_fct(model, model_type, descriptions, batch_size, mode='HF'):
    '''Fonction de création des features.'''
    bert_tokenizer = AutoTokenizer.from_pretrained(model_type)
    time1 = time.time()

    for step in range(len(descriptions)//batch_size) :
        idx = step*batch_size
        input_ids, token_type_ids, attention_mask, bert_inp_tot = bert_inp_fct(descriptions[idx:idx+batch_size], 
                                                                               bert_tokenizer)
        
        if mode=='HF' :    # Bert HuggingFace
            outputs = model.predict([input_ids, attention_mask, token_type_ids], batch_size=batch_size)
            last_hidden_states = outputs.last_hidden_state

        if mode=='TFhub' : # Bert Tensorflow Hub
            text_preprocessed = {"input_word_ids" : input_ids, 
                                 "input_mask" : attention_mask, 
                                 "input_type_ids" : token_type_ids}
            outputs = model(text_preprocessed)
            last_hidden_states = outputs['sequence_output']
             
        if step ==0 :
            last_hidden_states_tot = last_hidden_states
            last_hidden_states_tot_0 = last_hidden_states
        else :
            last_hidden_states_tot = np.concatenate((last_hidden_states_tot,last_hidden_states))
    
    features_bert = np.array(last_hidden_states_tot).mean(axis=1)
    
    time2 = np.round(time.time() - time1,0)
    print("temps traitement : ", time2)
     
    return features_bert, last_hidden_states_tot
```

```python
batch_size = 10
model_type = 'bert-base-uncased'
model = TFAutoModel.from_pretrained(model_type)
descriptions = df_cat['descr_dl'].to_list()
```

```python
# Création des features

features_bert, last_hidden_states_tot = feature_BERT_fct(model, model_type, descriptions, 
                                                        batch_size, mode='HF')
```

## <a id="5"></a> 1.4 USE: Universal Sentence Encoder

USE (Universal Sentence Encoder) is a natural language processing model developed by Google. It converts sentences into real-valued vectors, also known as embeddings, that capture the semantic meaning of the sentence. The model is trained on a large corpus of text in multiple languages and is capable of understanding the semantic and contextual nuances of sentences in different languages.


```python
def feature_USE_fct(descriptions, batch_size):
    time1 = time.time()

    for step in range(len(descriptions)//batch_size) :
        idx = step*batch_size
        feat = embed(descriptions[idx:idx+batch_size])

        if step ==0 :
            features = feat
        else :
            features = np.concatenate((features,feat))

    time2 = np.round(time.time() - time1,0)
    return features
```

# <a id="6"></a> 2. Visual Data Processing

We will now extract features from our visual data using a Transfer Learning algorithm based on neural networks such as CNN.

CNN stands for "Convolutional Neural Network." It is a type of artificial neural network widely used for image processing. They are designed to detect patterns in images using filters called "convolutions." These filters are applied to the input image to extract specific features such as edges, corners, textures, shapes, etc.

Our choice is oriented towards a VGG16 model. It is a deep convolutional neural network (CNN) with an architecture composed of 13 convolutional layers and 3 fully connected layers. The 13 convolutional layers are simple convolution blocks, each consisting of two or three convolutional layers with small filters (3x3) and a max pooling layer to reduce the dimensionality of the image. The three fully connected layers are used for final classification. 
It takes a color image of size 224x224 as input and classifies it into one of 1000 classes.

Here, we wish to use this pre-trained model not to classify our images into one of the 1000 classes but to extract features. To do this, we remove the last layers by adding **outputs=base_model.layers[-2].output**. Thus, we only retrieve the features.


```python
base_model = VGG16()
model = Model(inputs=base_model.inputs, outputs=base_model.layers[-2].output)

print(model.summary())
```
